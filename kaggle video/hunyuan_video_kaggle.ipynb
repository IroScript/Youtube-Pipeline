## Local Inference on GPU 
Model page: https://huggingface.co/tencent/HunyuanVideo

⚠️ If the generated code snippets do not work, please open an issue on either the [model repo](https://huggingface.co/tencent/HunyuanVideo)
			and/or on [huggingface.js](https://github.com/huggingface/huggingface.js/blob/main/packages/tasks/src/model-libraries-snippets.ts) 🙏

```python
!pip install -q diffusers transformers accelerate sentencepiece

import torch
from diffusers import HunyuanVideoPipeline
from diffusers.utils import export_to_video

# ১. মডেল লোড ([hunyuanvideo-community/HunyuanVideo](https://huggingface.co/tencent/HunyuanVideo))
pipe = HunyuanVideoPipeline.from_pretrained(
    "hunyuanvideo-community/HunyuanVideo",
    torch_dtype=torch.float16,
    device_map="balanced"
)

# ২. প্রম্পট
prompt = "A majestic lion standing on a mountain peak at sunset, realistic style, 4k"

# ৩. ভিডিও তৈরি ও সেভ
output = pipe(prompt=prompt, num_frames=61, num_inference_steps=30).frames[0]
export_to_video(output, "output_video.mp4", fps=15)
print("✅ ভিডিও তৈরি সম্পন্ন হয়েছে! 'output_video.mp4' ফাইলটি আউটপুটে জমা হয়েছে।")
```

```python
!pip install -q diffusers transformers accelerate sentencepiece

import torch
from diffusers import HunyuanVideoPipeline
from diffusers.utils import export_to_video

# ১. মডেল লোড
pipe = HunyuanVideoPipeline.from_pretrained(
    "hunyuanvideo-community/HunyuanVideo",
    torch_dtype=torch.float16,
    device_map="balanced"
)

# ২. VRAM বাঁচানোর জন্য CPU Offload চালু করা
pipe.enable_model_cpu_offload()

# ৩. প্রম্পট
prompt = "A majestic lion standing on a mountain peak at sunset, realistic style, 4k"

# ৪. ভিডিও তৈরি ও সেভ (মেমোরি বাঁচাতে num_frames=17 দেওয়া হয়েছে)
output = pipe(prompt=prompt, num_frames=17, num_inference_steps=20).frames[0]
export_to_video(output, "output_video.mp4", fps=15)
print("✅ ভিডিও তৈরি সম্পন্ন হয়েছে! 'output_video.mp4' ফাইলটি আউটপুটে জমা হয়েছে।")
```

```python
# ১. প্যাকেজ ইনস্টল (Kaggle নোটবুকের প্রথম সেলে আলাদা করে রান করুন)
# !pip install -q -U diffusers transformers accelerate bitsandbytes

import torch
from diffusers import HunyuanVideoPipeline, HunyuanVideoTransformer3DModel
from transformers import BitsAndBytesConfig
from diffusers.utils import export_to_video

# ২. 4-bit Quantization কনফিগারেশন
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_quant_type="nf4"
)

# ৩. Quantized Transformer লোড
transformer = HunyuanVideoTransformer3DModel.from_pretrained(
    "hunyuanvideo-community/HunyuanVideo",
    subfolder="transformer",
    quantization_config=quantization_config,
    torch_dtype=torch.float16,
    device_map="auto"
)

# ৪. পাইপলাইন লোড (device_map ছাড়া)
pipe = HunyuanVideoPipeline.from_pretrained(
    "hunyuanvideo-community/HunyuanVideo",
    transformer=transformer,
    torch_dtype=torch.float16,
)

# ৫. মেমোরি অপটিমাইজ
pipe.enable_model_cpu_offload()

# ৬. ভিডিও জেনারেট
prompt = "A majestic lion standing on a mountain peak at sunset, realistic style, 4k"
output = pipe(prompt=prompt, num_frames=17, num_inference_steps=20).frames[0]
export_to_video(output, "output_video.mp4", fps=15)
print("✅ 4-bit Quantized ভিডিও সফলভাবে তৈরি হয়েছে!")
```
